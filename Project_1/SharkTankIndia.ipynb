# For analyzing funding stage success, we don’t have a separate funding stage column, but we can analyze simple proxy using Investment Amount for funding stage.

```python
def funding_stage(amount):
    if amount == 0:
        return 'No Deal'
    elif amount < 10:
        return 'Seed'
    elif amount < 50:
        return 'Early'
    elif amount < 100:
        return 'Growth'
    else:
        return 'Expansion'

df['Funding Stage'] = df['Investment Amount (In Lakhs INR)'].apply(funding_stage)
```

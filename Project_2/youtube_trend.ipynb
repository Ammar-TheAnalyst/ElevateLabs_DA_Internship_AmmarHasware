# Converting boolean columns to 0/1, so that MySQL can accept them as integers

```python
bool_cols = ['comments_disabled', 'ratings_disabled', 'video_error_or_removed']
for col in bool_cols:
    df[col] = df[col].map({False: 0, True: 1, 'False': 0, 'True': 1})
```

```python
df.head(5)
```

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
```

# Creating sentiment and label columns for tags & title

```python
analyzer = SentimentIntensityAnalyzer()
df['title_sentiment'] = df['title'].apply(lambda x: analyzer.polarity_scores(x)['compound'])
df['tags_sentiment'] = df['tags'].apply(lambda x: analyzer.polarity_scores(x)['compound'])
```

```python
def sentiment_label(x):
    if x >= 0.05:
        return 'Positive'
    elif x <= -0.05:
        return 'Negative'
    else:
        return 'Neutral'

df['title_sentiment_label'] = df['title_sentiment'].apply(sentiment_label)
df['tags_sentiment_label'] = df['tags_sentiment'].apply(sentiment_label)
```
